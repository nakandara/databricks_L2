```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/dbacademy/default/tutorials/BigMart Sales.csv")
```

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
df.display()
```

###Date_Format()

```python
df = df.withColumn('curr_date',current_date()).display()
```

```python
df = df.withColumn('curr_date', date_format(current_date(),'dd-MM-yyyy'))

df.display()
```

### Handling Nulls


Dropping Nulls

```python
df.dropna('all').display()
```

```python
df.dropna('any').display()
```

```python
df.dropna(subset=['outlet_size']).display()
```

```python
df.display()
```

###FIlling Nulls

```python
df.fillna('NotAvailable').display()
```

```python
df.fillna('NotAvailable',subset=['Item_Weight']).display()
```
