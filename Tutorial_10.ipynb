# **User Defined Funtion(UDF)**

## Step 1

```python
def my_func(x):
  return x + 1
```

## Step 2

```python
my_udf = udf(my_func)
```

```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/dbacademy/default/tutorials/BigMart Sales.csv")
```

```python
df.withColumn('new_col', my_udf('Item_MRP')).display()
```

### Data Writing

```python
df.write.format('csv').save('/Volumes/dbacademy/default/tutorials/CSV/data.csv')
```

###APPEND

```python
df.write.format('csv')\
    .mode('append')\
    .option('path','/Volumes/dbacademy/default/tutorials/CSV/data.csv')\
    .save()
```

###Overwrite

```python
df.write.format('csv')\
    .mode('overwrite')\
    .option('path','/Volumes/dbacademy/default/tutorials/CSV/data.csv')\
    .save()
```

###Error

```python
df.write.format('csv')\
    .mode('error')\
    .option('path','/Volumes/dbacademy/default/tutorials/CSV/data.csv')\
    .save()
```

###Ignore

```python
df.write.format('csv')\
    .mode('ignore')\
    .option('path','/Volumes/dbacademy/default/tutorials/CSV/data.csv')\
    .save()
```

###Parquet

```python
df.write.format('parquet')\
.mode('overwrite')\
.option('path','/Volumes/dbacademy/default/tutorials/Parquet/data.parquet')\
.save()
```

**Table**

```python
df.write.format('parquet')\
    .mode('overwrite')\
    .saveAsTable('my_table')
```
