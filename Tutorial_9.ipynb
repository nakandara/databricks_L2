```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("/Volumes/dbacademy/default/tutorials/BigMart Sales.csv")
```

### Window Funtion

```python
from pyspark.sql.window import Window
```

#### row_number

```python
from pyspark.sql.functions import row_number

window = Window.orderBy('Item_Identifier')
df.withColumn('rowCol', row_number().over(window)).display()
```

#### Rank vs DenseRank

```python
from pyspark.sql.functions import rank

df.withColumn('rank', rank().over(window)).display()
```

```python
from pyspark.sql.functions import rank
from pyspark.sql.window import Window
from pyspark.sql.functions import col

windowdese = Window.orderBy(col('Item_Identifier').desc())
df.withColumn('rank', rank().over(windowdese)).display()
```

```python
from pyspark.sql.functions import rank, dense_rank
from pyspark.sql.window import Window
from pyspark.sql.functions import col

windowdese = Window.orderBy(col('Item_Identifier').desc())
df.withColumn('rank', rank().over(windowdese))\
    .withColumn('desentrank', dense_rank().over(windowdese)).display()
# The error was likely due to using col as a callable incorrectly elsewhere; this code is correct.
```

###Cumulatuve Sum

```python
from pyspark.sql.functions import sum
from pyspark.sql.window import Window

window_item_type = Window.orderBy('Item_Type')
df.withColumn('cumsum', sum('Item_MRP').over(window_item_type)).display()
```

```python
from pyspark.sql.functions import sum
from pyspark.sql.window import Window

window_item_type = Window.orderBy('Item_Type').rowsBetween(Window.unboundedPreceding, Window.currentRow)
df.withColumn('cumsum', sum('Item_MRP').over(window_item_type)).display()
```

```python
from pyspark.sql.functions import sum
from pyspark.sql.window import Window

window_item_type = Window.orderBy('Item_Type').rowsBetween(Window.unboundedPreceding, Window.unboundedFollowing)
df.withColumn('totalsum', sum('Item_MRP').over(window_item_type)).display()
```
